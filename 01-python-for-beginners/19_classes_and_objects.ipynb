### 1. Classes and Objects

**1. What is a class?**
- A class is a blueprint — like a design for building robots.

**2. What is an object?**
- An object is something created from that blueprint.

``Class: Robot``
- The design or template

``Object: robot1``
- A robot created from the design

**Python class format**

```python
class ClassName:
    def __init__(self, parameters):
        self.attribute = parameters

    def method_name(self):
        # Code to execute
        pass


# Create an object
object_name = ClassName(arguments)

# Call a method
object_name.method_name()
```

**Remember these terms**

| Term      | Simple meaning                     |
|-----------|------------------------------------|
| ``class``     | Blueprint                          |
| Object    | A thing created from the blueprint |
| Attribute | Information about the object /  store information.     |
| Method    | An action the object can perform   |
| ``__init__() ``|initializes the object.|
| ``self`` | refers to the current object.|

**3. Let's write the code**

* Let’s create a simple class called Robot. 

* This class will have some basic attributes like name and color, and a method that allows the robot to introduce itself.

```python
class Robot:
    def __init__(self, name, color):
        self.name = name
        self.color = color
 
    def introduce_self(self):
        print(f"Hello, my name is {self.name} and my color is {self.color}.")
```

Now, let’s create an object of the Robot class:

```python
robot1 = Robot("Atlas", "red")
robot2 = Robot("Spot", "blue")

robot1.introduce_self()
robot2.introduce_self()
```

**2. With __ — private-style**

```python
class Robot:
    def __init__(self, name):
        self.__name = name

robot = Robot("R2D2")

print(robot.__name)
# AttributeError: 'Robot' object has no attribute '__name'
```

To access it, you can create a method:

```python
class Robot:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

robot = Robot("R2D2")

print(robot.get_name())
```

**Example**

```python
class Robot:
    # Global class variables can be defined here
    everyRobotHasThis = 'All robots count'

    # Constructor method
    def __init__(self, name, howCool='very cool'):
        # Private Variable
        self.__name = name
        self.__howCool = howCool

    # Method to perform a cool action
    def doSomethingCool(self):
        print(f"{self.__name} lift right arm")

    # Static method to indicate the robot is too cool to instantiate
    @staticmethod
    def toCoolToInstant():
        print("I am special. Who needs an object?")

    def printName(self):
        print(self.__name)

    # setter and getter for name
    # To change the name of the robot
    # newName: the new name to assign to the robot
    def setName(self, newName):
        self.__name = newName

my_robot = Robot("R2D2")
print(my_robot._Robot__name)
print(my_robot._Robot__howCool)

my_robot.doSomethingCool()

# Static method call
Robot.toCoolToInstant()

# Accessing and modifying the private variable directly
my_robot.printName()
my_robot._Robot__name= "C3PO"
my_robot.printName()

# Using the setter method to change the name
my_robot.setName("R2D3")
my_robot.printName()
print(my_robot.everyRobotHasThis)

my_robot2 = Robot("BB-8")
print(my_robot2.everyRobotHasThis)

debug = 1
```
