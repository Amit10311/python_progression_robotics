### Lists

**What is a list?**
- A list is a collection of items stored in one variable.
- List is **object**, it contains **methods**. 

```python
fruits = ["apple", "banana", "mango"]
```

A list can hold:
- strings
- numbers
- booleans
- other lists

```python
list[start : end]
```
- start is included
- end is excluded


### 1. Creating a list


````python
numbers = [1, 2, 3, 4, 5]
names = ["Amit", "Riya", "Ali"]
````


### 2. Accessing elements
Each item in a list has an index.

````python
fruits = ["apple", "banana", "mango"]

print(fruits[0])  # apple
print(fruits[1])  # banana
print(fruits[-1]) # mango
````

### 3. Changing values

- Lists are mutable, which means you can change them.

````python
fruits = ["apple", "banana", "mango"]
fruits[1] = "grape"

print(fruits)   # ['apple', 'grape', 'mango']
````

### 4. Adding items
#### 4.1. **`append()`**

````python
fruits.append("orange")
print(fruits)
````

#### 4.2. `insert()`

````python
fruits.insert(1, "kiwi")
print(fruits)
````

### 5. Removing items
#### 5.1. `remove()`

````python
fruits.remove("banana")

print(fruits)
````

#### 5.2. `pop()`

````python
fruits.pop()   # removes last item

print(fruits)
````

### 6. Length of list

````python
print(len(fruits))
````

### 7. Looping through a list


````python
for item in fruits:
    print(item)
````

### 8. Slicing


````python
numbers = [10, 20, 30, 40, 50]

index:     0    1    2    3    4
value:    10   20   30   40   50
````

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])   # [20, 30, 40]
    # - start from index 1
    # - stop before index 4

print(numbers[:3])    # [10, 20, 30]
    # - start from index 0
    # - stop before index 3

print(numbers[2:])    # [30, 40, 50]
    # - start from index 2 to end
````


### 9 List methods
Here are some common ones:

````python
numbers = [3, 1, 2]

numbers.sort()
print(numbers)   # [1, 2, 3]

numbers.reverse()
print(numbers)   # [3, 2, 1]

numbers.count(2)  # returns how many times 2 appears

numbers.append(2)
print(numbers)   # [3, 2, 1,1]

print(len(numbers)) 
numbers.insert(0,"a")
print(numbers) 
````

#### Summary
A list is:
- ordered
- changeable
- indexed
- useful to store multiple values

````python
# 1. Create a list
names = ["John", "Bob", "Mosh", "Sam", "Mary"]
print("Original list:", names)

# 2. Accessing elements
print("First name:", names[0])
print("Last name:", names[-1])
print("Second-last name:", names[-2])

# 3. Changing values
names[0] = "Jon"
print("After changing the first value:", names)

# 4. Adding items
names.append("Sara")
print("After append:", names)

names.insert(1, "Amit")
print("After insert Amit :", names)

# 5. Removing items
names.remove("Sam")
print("After remove Sam:", names)

popped_name = names.pop()
print("After pop:", names)
print("Removed item:", popped_name)

# 6. Length of the list
print("Length of the list:", len(names))

# 7. for Loop through the list
for name in names:
    print(name)

# 8. Slicing
print("First 3 items:", names[0:3])
    # means start from beginning to index 3, stop before index 3

# 9. Membership check
print("Is 'Amit' in the list?", "Amit" in names)
print("Is 'Sara' in the list?", "Sara" in names)

# 10. Clear all items
names.clear()
print("After clearing:", names)
````
