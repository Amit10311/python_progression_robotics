## While loops

* A while loop repeats a block of code as long as a condition is True.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### Structure
```python
while condition:
    # code to repeat
```

- Condition is checked before every iteration.
- If condition is `True`, loop runs.
- If condition becomes `False`, loop stops.


#### Important
If the condition never becomes `False`, the loop becomes infinite.

```python
while True:
    print("This will run forever")
```

- This is called an **infinite loop**.

````python
num = 1

while num <= 10:
    print(num)
    num += 1
````

#### 2. `break` statement
- `break` stops the loop immediately.

````python
count = 1

while count <= 10:
    print(count)
    if count == 5:
        break
    count += 1
````

#### 3. `continue` statement

- `continue` skips the rest of the loop and goes to the next iteration.

````python
count = 0

while count < 5:
    count += 1
    if count == 3:
        continue
    print(count)
````

#### Real life example
A `while` loop is useful when you don’t know in advance how many times to repeat.

**Example:** keep asking until user enters correct password.


````python
password = ""

while password != "python":
    password = input("Enter password: ")
    if password != "python":
        print("Wrong password, try again")
````

**Example**

````python
i = 1

while i <= 10:
    print (i * '*')
    i = i +1
````

````python
i = 1

while i <= 10:
    print (str(i) + '*')
    i = i +1
````

#### Summary

- `while` = repeat while condition is true
- Must change condition to eventually stop
- Use `break` to stop early
- Use `continue` to skip an iteration
