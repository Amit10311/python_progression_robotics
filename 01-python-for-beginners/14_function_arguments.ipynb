### 2. Function Arguments 

**Parameter**
- A parameter is a variable declared inside a function definition.

````python
def greet(name):
    print("Hello", name)
````

**Argument**
- An argument is the actual value passed to the function when calling it.

````python
greet("Amit")
````

````python
def function(parameter1, parameter2):
    ...
function(argument1, argument2)
````

````python
def greet():
    print("Hello!") # This function takes arguments
    print("How are you?")

greet()  # This function is called without any arguments
````

````python
def greet(first_name, last_name):
    print("Hello!") # This function takes arguments
    print("How are you?")

greet("Amit", "Sharma")  # This function is called with two arguments
````

### 3. Types of Functions

**1. Functions with Parameters**

- Functions that receive values.

```python
def add(a, b):
    return a + b
```

**2. Functions with Default Arguments**
- Arguments that already have a default value.

```python
def greet(name="User"):
    print("Hello", name)
```

**3. Functions with Return Value**
- Functions that return a result.

```python
def square(x):
    return x * x
```

**4. Functions without Return Value**
- Functions that only print or do an action.

```python
def show():
    print("This is a function")
```

**5. Recursive Functions**
- A function that calls itself.

```python
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)
```


### 4. Keyword Arguments

A keyword argument is an argument passed to a function using the parameter name.

````python
def increment(number, by):
    return number + by

result = increment(5, 2)
print(result)

print(increment(5, by=3))

def greet(name, age):
    print("Name:", name)
    print("Age:", age)

greet(name="Amit", age=25)
greet(age=25, name="Sunil")  

# greet(age=25, n="Amit")   # error, because n is not a valid parameter
````

### 5. Default Arguments

- A default argument is a parameter that already has a value assigned in the function definition.

- If the caller does not pass that argument, Python uses the default value automatically.

````python
def greet(name="User"):
    print("Hello,", name)

greet()

greet("Amit")
````

````python
def area(length, width=10):
    return length * width

print(area(5))      # 50
print(area(5, 4))   # 20
````

**Important**
Default arguments must come after non-default arguments.

✅ Correct:
```python
def my_func(a, b=5):
    ...
```

❌ Incorrect:
```python
def my_func(a=5, b):
    ...
```


### 6. xargs

````python
def multiply(a, b=2):
    return a * b

multiply(5, 3, 5, 2 )  # error, too many arguments provided
````

````python
def multiply(*numbers):
    print(numbers)

multiply(5, 3, 5, 2 )  # error, too many arguments provided
````

````python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

print(multiply(5, 3, 5, 2))  # 150
````
