## ``range()`` functions

#### What is `range()`?
`range()` generates a sequence of numbers.

It is commonly used in `for` loops.


#### Syntax
```python
range(start, stop, step)
```

- `start` = first number (included)
- `stop` = last number (not included)
- `step` = difference between numbers (default is 1)

---

### Examples


#### 1. `range(5)`

````python
for i in range(5):
    print(i)
````

#### 2. `range(2, 7)`

````python
for i in range(2, 7):
    print(i)
````

#### 3. `range(1, 10, 2)`

````python
for i in range(1, 10, 2):
    print(i)
````

#### Important point
- `range()` stops before the final value.

**Example:**

```python
range(5)
```
- means:
```python
0, 1, 2, 3, 4
```
- not 5.

#### 4. Using `range()` with list

````python
numbers = list(range(1, 6))
print(numbers)
````


#### Summary

- `range()` helps create number sequences quickly, especially in loops.
- `range(stop)` → starts at 0
- `range(start, stop)` → starts at start
- `range(start, stop, step)` → uses step value

**Examples**

1) Print numbers from 0 to 4

````python
for i in range(5):
    print(i)
````

2) Print numbers from 2 to 5

````python
for i in range(2,6):
    print(i)
````

3) Print odd numbers from 1 to 9


````python
for i in range(1,10,2):
    print(i)
````

4) Print numbers in reverse from 5 to 1

````python
for i in range(5,0,-1):
    print(i)
````

5) Create a list from range

````python
numbers = list(range(1,6))
print(numbers)
````

6) Print every 3rd number from 0 to 15


````python
for i in range(0,16,3):
    print(i)
````

7) Print numbers from 20 down to 5


````python
for i in range(20, 4, -3):
    print(i)
````

8) Print even numbers from 10 to 0


````python
for i in range(10, -1, -2):
    print(i)
````

9) Multiply numbers by 10


````python
for i in range(1,6):
    print(i*10)
````

10) Create a list of squares


````python
squares = [x**2 for x in range(1,7)]
print(squares)
````

11) Function to return squares

````python
def square_list(n):
    result = []
    for i in range(1, n + 1):
        result.append(i * i)
    return result

print(square_list(5))
````
