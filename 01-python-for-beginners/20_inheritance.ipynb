## Python Inheritance

Inheritance means a class can reuse attributes and methods from another class.

1. **Base class (parent) :** The class whose features are inherited.

2. **Derived class (child) :** The class that inherits features from the base class.

**Example: Robotics 🤖**

```python
class Robot:
    def move(self):
        print("Robot is moving")


class Drone(Robot):
    def fly(self):
        print("Drone is flying")

# Create an instance of Drone and demonstrate inheritance and its own method
# Create a Drone object 
drone = Drone()
# Demonstrate inheritance and its own method
drone.move()  # Inherited from Robot
drone.fly()   # Defined in Drone
```

**Remember:** ``class Drone(Robot)`` means ``Drone`` inherits from Robot.

```python
class Robot:
    def __init__(self, name, id):
        self.__name = name
        self.__id = id

    def printName(self):
        print("my name is :", self.__name)
        print("my id is :", self.__id)

class CoolRobot(Robot):
    def __init__(self, name, id, coolnessLevel):
        # Initialize the parent class with name and id
        super().__init__(name, id)
        self.__coolnessLevel = coolnessLevel

myCoolRobot = CoolRobot("Amit", 10, 1001)
myCoolRobot.printName()


debug = 1
```

#### Abstract Method 

```python
class Robot:
    def __init__(self, name, id):
        self.__name = name
        self.__id = id

    def printName(self):
        print("my name is :", self.__name)
    
    def doSomething(self):
        return "Doing something"

# Derived Class 1 
class CoolRobot(Robot):
    def __init__(self, name, id, coolnessLevel):
        # Initialize the parent class with name and id
        super().__init__(name, id)
        self.__coolnessLevel = coolnessLevel

    def doSomething(self):
        print(self._Robot__name, 'raise eyebrow')

# Derived Class 2 
class BoaringRobot(Robot):
    def __init__(self, name, id):
        # Initialize the parent class with name and id
        super().__init__(name, id)

    def doSomething(self):
        print(self._Robot__name, 'zzzzzz')



myCoolRobot = CoolRobot("Amit", 10, 1001)
myCoolRobot.printName()

# Abstract Class Example
myCoolRobot.doSomething()

myBoaringRobot = BoaringRobot("Bob", 20)
myBoaringRobot.printName()
myBoaringRobot.doSomething()


debug = 1
```
