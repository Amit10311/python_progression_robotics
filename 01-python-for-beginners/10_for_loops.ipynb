## For Loops

**What is a for loop?**
- A `for` loop is used to repeat a block of code for each item in a sequence.

Examples of sequences:
- list
- string
- tuple
- range


**syntax**
```python
for item in sequence:
    print(item)
```

##### Example 1: looping through a list

````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
````

##### Example 2: looping through a string

````python
word = "Python"

for ch in word:
    print(ch)
````

##### Example 3: using `range()`

````python
for i in range(5):
    print(i)
````

#### 2. `range(start, stop, step)`

````python
for i in range(1, 10, 2):
    print(i)
````

#### 3. Using `for` with `if`


````python
numbers = [1, 2, 3, 4, 5]

# Print the even numbers from the lists 
print(len(numbers) != 0, "\nLength of the list:", len(numbers))

print("Checking even numbers...")
for num in numbers:
    if num % 2 == 0:
        print(num)

print("\nCheck all numbers")
i = 0

while i < len(numbers):
    print(numbers[i])
    i = i +1 
````

#### Difference between `for` and `while`

- `for` loop is used when you know how many times or have a collection
- `while` loop is used when condition decides the repetition

##### Summary
A `for` loop:
- goes through each item in a sequence
- repeats code automatically
- is best for lists, strings, and ranges

````python
successful = True

for num in range(5): 
    if successful:
        print(f"Attempt {num} was successful.")
    else:
        print(f"Attempt {num} failed.")
````

````python
successful = True

for num in range(5): 
    if successful:
        print(f"Attempt {num} was successful.")
        break
````
