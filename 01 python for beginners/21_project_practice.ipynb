**EXERCISE 2 : Calculator**

- First number: 10.1
- Second number: 20
- Calculate the sum

```python
first_num = input("First: ")
# first_num = "10.1"  
second_num = input("Second: ")
# second_num = "20"

result = float(first_num) + float(second_num)

print("Sum : "+ str(result))
print(type(result))
```

**📝 Mistakes — Python Basics**
- Avoid using ``sum`` as a variable name because Python has a built-in function called ``sum()``.
- Python automatically converts the ``int`` to a ``float`` when needed but not when to take input.
