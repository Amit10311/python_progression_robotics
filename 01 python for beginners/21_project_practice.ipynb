**⚖️ Exercise 3 — Weight Conversion**

**Task:** Create a program that converts weight between **kilograms (kg)** and **pounds (lbs)**.

**Example:**
- Enter weight: `170`
- Enter unit: `L` for pounds
- Convert the weight to kilograms
- Expected result: `76.5 kg`

**Formula used in the exercise:**

```bash
kg = lbs × 0.45
```


````python
weight = float(input("Weight : "))
unit = input("(K)g or (L)bs: ")

if unit.upper() == "L":
    converted = weight * 0.45
    print("Weight in Kg:", converted)
elif unit.upper() == "K":
    converted = weight / 0.45
    print("Weight in Lbs:", converted)
else:
    print("Invalid unit")
````

**Exercise 5 - Factorial**

- Factorial means multiplying a number by every positive integer below it.

````python
def factorial(n):
    result = 1
    
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))
````

**Exercise 6 - Prime Numaber**

- Check if a number is prime

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(17))  # True
print(is_prime(20))  # False
````

**Exercise 7 - Fibonacci sequence**

````python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        print(a, end=" ")
        a, b = b, a + b

fibonacci(8)
````

**Exercise 8 -  Find the largest number in a list**

````python
def largest_number(nums):
    largest = nums[0]
    for num in nums[1:]:
        if num > largest:
            largest = num
    return largest

print(largest_number([12, 45, 7, 89, 2]))  # 89
````

**Exercise 9 -  Find the largest number in a list**

````python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

print(count_vowels("Hello World"))  # 3
````
