### Tuples

**What is a tuple?**
- A tuple is an ordered collection of values, like a list, but it is **immutable**.

That means:
- you cannot change its items after creation
- you cannot add or remove items


### Syntax


```python
numbers = (1, 2, 3, 4)
```

##### Example

```python
student = ("Amit", 25, "Python")
print(student)
```


### 1. Accessing values

- Like lists, tuple items are accessed by index.

```python
student = ("Amit", 25, "Python")

print(student[0])  # Amit
print(student[1])  # 25
```


### 2. Tuple is immutable

```python
student = ("Amit", 25, "Python")

# student[1] = 30   # This will raise an error
```

### 3. Length of tuple

```python
print(len(student))
```


### 4. Tuple unpacking
You can assign tuple values to variables.

```python
name, age, course = ("Amit", 25, "Python")
print(name)    # Amit
print(age)     # 25
print(course)  # Python
```


### 5. Creating a tuple with one value

- Need a comma:

```python
x = (5,)
print(type(x))
```

- Without comma:

```python
x = (5)
print(type(x))
```

This is just an integer, not a tuple.

### 6. Common use
Tuples are often used when values should not change, such as:
- coordinates
- date values
- fixed records

### 7. Comparison with list
| Type | Mutable | Example |
|---|---|---|
| List | Yes | `[1, 2, 3]` |
| Tuple | No | `(1, 2, 3)` |
