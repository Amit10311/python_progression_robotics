### 1. Arithmetic Operators 


| Operator | Name | Example (`a=10`, `b=3`) | Result |
|---|---|---|---|
| `+` | Addition | `a + b` | `13` |
| `-` | Subtraction | `a - b` | `7` |
| `*` | Multiplication | `a * b` | `30` |
| `/` | Division | `a / b` | `3.3333...` |
| `//` | Floor Division | `a // b` | `3` |
| `%` | Modulus (Remainder) | `a % b` | `1` |
| `**` | Exponent (Power) | `a ** b` | `1000` |

---
#### Quick code


````python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.3333...
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
````

### 2. Augmented Assignment Operators (Python)

These operators combine an operation + assignment in one step.

Example:

```python
x = 10
x += 5   # same as: x = x + 5
print(x) # 15
```
---
#### Common augmented operators

| Operator | Meaning | Same as |
|---|---|---|
| `+=` | Add and assign | `x = x + y` |
| `-=` | Subtract and assign | `x = x - y` |
| `*=` | Multiply and assign | `x = x * y` |
| `/=` | Divide and assign | `x = x / y` |
| `//=` | Floor divide and assign | `x = x // y` |
| `%=` | Remainder, Modulus and assign | `x = x % y` |
| `**=` | Power and assign | `x = x ** y` |

---

**Example code**

````python
x = 10

x += 2   # 12
x -= 3   # 9
x *= 4   # 36
x /= 6   # 6.0
x //= 2  # 3.0
x %= 2   # 1.0
x **= 3  # 1.0
````

````python
x = 10
print (x)
print ("Divide:", x/3)
print ("Floor Divide:", x//3)
print ("Modulus/Remainder :", x%3)

x %= 2   # 1.0s
print (x)
````

**With strings too**

````python
text = "Hi"
text += " Python"
print(text)   # Hi Python
````

### 3. Operator Precedence 

* **Operator precedence** means: which operator runs first in an expression.

#### **Order (high → low)**

1. `()` Parentheses  
2. `**` Exponent  
3. `+x`, `-x` Unary plus/minus  
4. `*`, `/`, `//`, `%`  
5. `+`, `-`  

If same level, Python usually evaluates **left to right**  
(except `**`, which is right to left).

---

**Examples**


````python
print(2 + 3 * 4)      
print((2 + 3) * 4)    
print(2 ** 3 ** 2)    
print(20 / 5 * 2)     
print(20 // 3 + 2)    
````

````python
print(2 + 3 * 4)        # 14   (* first)
print((2 + 3) * 4)      # 20   (parentheses first)
print(2 ** 3 ** 2)      # 512  (2 ** (3 ** 2))   = 2 ^ 9 
print(20 / 5 * 2)       # 8.0  (left to right)
print(20 // 3 + 2)      # 8
````

##### **Easy rule to remember**
Use **BODMAS/PEMDAS style** thinking:
- Brackets first
- Powers
- Multiply/Divide
- Add/Subtract
