*   A variable is a name used to store data in memory.

```python
# Variables 
age = 20
price = 19.99

# String Types 
first_name ="Mosj"


print(age)
print(price)
```

### **2. Rules for Variable**

✅ Allowed:
* letters, numbers, underscore (_)
* must start with letter or underscore

❌ Not allowed:
* cannot start with number
* no spaces
* no special symbols like @, #, %

Examples:

```python
student_name = "Ravi"   # valid
_age = 20               # valid

2name = "test"          # invalid
student name = "Ravi"   # invalid
```

### 3. Main Python Data Types

* Whole numbers:


```python
x = 10
y = -5
```

b. ``float`` 

```python
price = 99.99
pi = 3.14
```

c. ``str`` (String)


```python
city = "Delhi"
letter = 'A'
```

d. ``bool`` (Boolean)

* Only ``True`` or ``False``:

```python
fruits = ["apple", "banana", "mango"]
```

### 4. Check Data Type using ``type()``


```python
a = 10
b = 10.5
c = "hello"
d = True

print(type(a))  # <class 'int'>
print(type(b))  # <class 'float'>
print(type(c))  # <class 'str'>
print(type(d))  # <class 'bool'>
```
