# ``if`` Condition


**1. if statement**


```python
temperature = 35 

if temperature > 30 : 
    print("It is a hot day")
    print("Drink plenty of the water")
```

**2. else statement**


```python
temperature = 25

if temperature > 30 :   # (30, +inf)
    print("It is a hot day")
    print("Drink plenty of the water")
else:                   # (-inf, 30)
    print("It's cold")
print("Done")
```

3. **elif statement**

- Used when there are multiple conditions.

```python
temperature = 10

if temperature > 30 :   # (30, +inf)
    print("It is a hot day")
    print("Drink plenty of the water")
elif temperature > 20 : # (20, 30]
    print("It's a nice day")
elif temperature > 10:  # (10, 20]
    print("It's a cold day")
else:                   # (-inf, 10]
    print("It's cold")
print("Done")
```

**4. Nested if**

```python
age = 22
is_student = True

if age >= 18:
    if is_student:
        print("Adult student")
    else:
        print("Adult not student")
else:
    print("Minor")
```

### 2. Chaining Comparison Operators

```python
age = 22

if 18 <= age < 65:
    print("Eligible")
```

```python
if 10 == '10':
    print("a")
elif "bag" > "apple" and "bag" > "cat":
    print("b")  
else:
    print("c")
```

**Explanation**

Python compares strings by letter order:

- ``"a" < "b" < "c" ...``
- ``"apple"`` comes before ``"bag"``

```python
"bag" > "apple"   # True
```

```python
"bag" > "cat"   # False
```
