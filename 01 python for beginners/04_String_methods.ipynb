### 1. Function
A **function** is a reusable block of code.

- `def` creates function
- `name` is parameter
- `return` sends result back

```python
def greet(name):
    return f"Hello, {name}!"
```

* Use it: 

```python
print(greet("Amit"))
```

- A **class** is a blueprint for creating objects.

```python
class Student:
    pass
```

```python
class Student:
    school_name = "ABC School"   # class attribute
```

### 4. Instance Attributes
* An **instance attribute** belongs to each object separately.

- Variable = general word
- Attribute = variable that belongs to a class/object

```python
class Student:
    school_name = "ABC School"   # class attribute

    def __init__(self, name, marks):
        self.name = name         # instance attribute
        self.marks = marks       # instance attribute
```

### 5. Methods (Functions inside a class)
* A method is just a function inside a class.

```python
class Student:
    school_name = "ABC School"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):   # instance method
        print(f"Name: {self.name}, Marks: {self.marks}")
```

### 6) “Main method” in Python

#### Why use this?

- Code inside `if __name__ == "__main__":` runs only when file is run directly.
- It does not run when file is imported in another file.

```python
def main():
    s1 = Student("Amit", 90)
    s1.display()
    print("School:", Student.school_name)

if __name__ == "__main__":
    main()
```

### Full Example (all together)

```python
class Student:
    # school_name = "ABC School"   # class attribute

    def __init__(self, name, marks):
        self.name = name         # instance attribute
        self.marks = marks       # instance attribute

    def display(self):           # method
        print(f"Name: {self.name}, Marks: {self.marks}")

def main():
    Student.school_name = "XYZ School"   # class attribute
    s1 = Student("Amit", 90)
    s2 = Student("Riya", 85)

    s1.display()
    s2.display()

    print("School:", Student.school_name)

if __name__ == "__main__":
    main()
```

# String Objects 

- ``str`` is the built-in Python class for text.


```python
a = "hello"
b = 'python'
print(type(a))   # <class 'str'>
```

### **Important properties of str object**


1. Immutable (cannot change directly)


```python
s = "cat"
# s[0] = "b"   # Error
s = "bat"       # create new string
```

2. Ordered sequence of characters


```python
s = "python"
print(s[0])   # p
print(s[-1])  # n
```

3. Supports slicing


```python
s = "python"
print(s[0:3])  # pyt
```

**Common str methods**

```python
text = "  Hello Python  "

print(text.lower())      # "  hello python  "
print(text.upper())      # "  HELLO PYTHON  "
print(text.strip())      # "Hello Python"

print(text.replace("Python", "World"))  # "  Hello World  "

print("Length of the String ", len(text))   # length of string

print( "index position :", text.find("Python"))  # index position , gives of the first character 

print(text.split())      # ['Hello', 'Python']
```
