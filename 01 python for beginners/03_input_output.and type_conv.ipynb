# Input /Output  and Type Conversions

### 1) Output in Python: ``print()``

Use ``print()`` to display text or values.

* Print multiple values


```python
name = "Amit"
age = 25
print("Name:", name, "Age:", age)
```

* ``sep`` and ``end``

```python
print("2026", "10", "04", sep="-")   # 2026-10-04
print("Hello", end=" ")
print("World")                        # Hello World
```

### 2) Input in Python: ``input()``


```python
name = input("What is your name ")

# String concatenation 
print ("Hello " + name)
```

**⚠ Important: ``input()`` always returns ``str``.**


### 3) Convert input to other types

1. ``int()`` input

```python
birth_year = input("Enter your birth date ")
age = 2026 - birth_year 
# age = 2020 - "1996" -> We need to convert in into int 
print(age)
```

```python
birth_year = input("Enter your birth date ")
age = 2026 - int(birth_year )
print(age)
```

Example : 

```python
age = int(input("Enter your age: "))
print("Next year age:", age + 1)
```

2. ``float`` input

```python
price = float(input("Enter price: "))
print("Price with tax:", price * 1.18)
```

### Common beginner mistakes

1. Forgetting type conversion:


```python
age = input("Age: ")
print(age + 1)   # error
```

Fix:



```python
age = int(input("Age: "))
print(age + 1)   
```

2. Invalid number input:


```python
num = int(input("Enter number: "))  # crashes if user types abc
```

**Safer version:**

```python
try:
    num = int(input("Enter number: "))
    print(num)
except ValueError:
    print("Please enter a valid integer.")
```
