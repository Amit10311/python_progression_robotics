## 1. Comparison Operators

**What are Comparison Operators?**

Comparison operators compare two values and return:
- `True` or
- `False`

---

### 1.1 List of Comparison Operators

| Operator | Meaning  | Example  | Result |
|-------|-------------|----------|--------|
| `==` | Equal to     | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>`  | Greater than | `7 > 4`  | `True` |
| `<`  | Less than    | `2 < 1`  | `False` |
| `>=` | Greater than or equal to | `6 >= 6` | `True` |
| `<=` | Less than or equal to    | `3 <= 2` | `False` |

---

**Example Code**


```python
a = 10
b = 5

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a < b)    # False
print(a >= 10)  # True
print(b <= 5)   # True
```

### 1.2. With Strings

```python
print("apple" == "apple")   # True
print("cat" > "bat")        # True (alphabetical comparison)
```

### 1.3.  In `if` conditions


```python
age = 18

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible")
```

### 2. Logical Operators

There are 3 logical operators:

1. `and`
2. `or`
3. `not`

#### 2.1. `and`
- Returns `True` only if **both** conditions are true.

```python
age = 20
has_id = True

print(age >= 18 and has_id)   # True
```

```python
price = 25

print(price > 10 and price < 30)
```


#### 2.2. `or`

- Returns `True` if **at least one** condition is true.

```python
is_weekend = False
is_holiday = True

print(is_weekend or is_holiday)   # True
```

```python
price = 5 

print(price > 10 or price < 30)
```


### 2.3. `not`
Reverses the result:
- `True` → `False`
- `False` → `True`

```python
is_raining = False
print(not is_raining)   # True
```

#### Truth idea (simple)


| Expression | Result |
|---|---|
| `True and True` | `True` |
| `True and False` | `False` |
| `True or False` | `True` |
| `False or False` | `False` |
| `not True` | `False` |


**Combined example**

```python
marks = 75
attendance = 85

if marks >= 40 and attendance >= 75:
    print("Pass")
else:
    print("Fail")
```

**Real-life example**

```python
username = "admin"
password = "1234"

if username == "admin" and password == "1234":
    print("Login successful")
else:
    print("Invalid credentials")
```

### Important Note
- Use `==` for comparison
- Use `=` for assignment

```python
x = 5      # assignment
x == 5     # comparison
```

#### 3. Ternary Operator

```python
age = 18

if age >= 18:
    message = "Eligible"
else:
    message = "Not Eligible"
print(message)
```

```python
message = "Eligible" if age >= 18 else "Not Eligible"
print(message)
```

**Example**

```python
high_income = False
good_credit = True 

if high_income and good_credit:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
high_income = False
good_credit = True 

if high_income or good_credit:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
high_income = False
good_credit = True 
student = False

if not student:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
high_income = False
good_credit = True 
student = False

if (high_income or good_credit) and not student:
    print("Eligible")
else:
    print("Not Eligible")
```

### 4. Short-circuit Evaluations

Short-circuit evaluation saves time and prevents unnecessary work.

- ``and``: stop when first ``False``
- ``or``: stop when first ``True``


```python
print(False and (10 / 0))   # No error, because it stops early
```

```python
print(True or (10 / 0))     # No error, because it stops early
```

```python
x = 0
print(x != 0 and 10 / x)   # stops early because x != 0 is False
```

**Real-life use**


```python
name = ""
print(name and "Hello")    # prints empty string
```

```python
name = "Amit"
print(name and "Hello")    # prints Hello
```

**Because:**

- empty string is falsey
- non-empty string is truthy
